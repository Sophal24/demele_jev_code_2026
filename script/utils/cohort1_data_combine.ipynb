```python
# Import
import pandas as pd 
```

```python
# Path

csv_cohort1_a1 = '../../dataset/cohort1/A1_Questionnaire_Cohort1_Baseline_data.csv'
csv_cohort1_b1 = '../../dataset/cohort1/B1_Sample_Collection_Form_Cohort1_Baseline_data.csv'
csv_cohort1_c1 = '../../dataset/cohort1/C1_LabResult_Cohort1_Baseline_ELISA_data.csv'
```

```python
df_a1 = pd.read_csv(csv_cohort1_a1, low_memory=False)
df_b1 = pd.read_csv(csv_cohort1_b1, low_memory=False)
df_c1 = pd.read_csv(csv_cohort1_c1, low_memory=False)

df_a1.head()
```

```python
# df_a1[df_a1['A1'].map(df_a1['A1'].value_counts()) == 2].head()
# df_a1[df_a1['A1'] == 'DEM1-SR-0780'].head() 
```

```python
def make_list_by_id(df):
    df = df.copy()
    manual_id = df['A1manual'].notna()
    df.loc[manual_id, 'A1'] = df.loc[manual_id, 'A1manual']

    return (
        df
        .rename(columns={'A1': 'ID'})
        .groupby('ID', as_index=False)
        .agg(lambda col: col.dropna().tolist())
    )


def prefix_columns(df, prefix):
    return df.rename(columns={col: f'{prefix}_{col}' for col in df.columns if col != 'ID'})


df_a1_list = make_list_by_id(df_a1)
df_b1_list = make_list_by_id(df_b1)
df_c1_list = make_list_by_id(df_c1)

df_cohort1 = (
    prefix_columns(df_a1_list, 'a1')
    .merge(prefix_columns(df_b1_list, 'b1'), on='ID', how='outer')
    .merge(prefix_columns(df_c1_list, 'c1'), on='ID', how='outer')
)

df_cohort1.head()
```
